```python
# 1.Additive operation
def calculator():
    print("=== Simple Calculator ===")

    # 1. Prompt for first number
    try:
        num1 = float(input("Enter first number: "))
    except ValueError:
        print("Invalid input! Please enter a valid number.")
        return

    # 2. Choose operator
    print("\nSelect an operator:")
    print(" + : Addition ")
    print(" - : Subtraction ")
    print(" * : Multiplication")
    print(" / : Division ")
    operator = input("Enter operator (+, -, *, /): ").strip()

    # 3. Prompt for second number
    try:
        num2 = float(input("Enter second number: "))
    except ValueError:
        print("Invalid input! Please enter a valid number.")
        return

    # 4. Process and display clear math result
    print("\n" + "=" * 25)
    if operator == '+':
        result = num1 + num2
        print(f"Result: {num1} + {num2} = {result}")
    elif operator == '-':
        result = num1 - num2
        print(f"Result: {num1} - {num2} = {result}")
    elif operator == '*':
        result = num1 * num2
        print(f"Result: {num1} × {num2} = {result}")
    elif operator == '/':
        if num2 == 0:
            print("Error: Division by zero is not allowed.")
        else:
            result = num1 / num2
            print(f"Result: {num1} ÷ {num2} = {result}")
    else:
        print("Invalid operator selected.")
    print("=" * 25)

if __name__ == "__main__":
    calculator()
```
