```python
def qa_bot():
    print("=== Welcome to the QA Bot! ===")
    print("Type your question below (or type 'exit' to quit).\n")

    while True:
        # Prompt for user input and convert it to lowercase for case-insensitive matching
        user_input = input("You: ").strip().lower()

        # Check if the user wants to exit
        if user_input == "exit":
            print("Bot: Goodbye! Have a great day!")
            break

        # Keyword 1: 'hello' or 'hi'
        if "hello" in user_input or "hi" in user_input:
            print("Bot: Hello there! How can I help you today?")

        # Keyword 2: 'python'
        elif "python" in user_input:
            print("Bot: Python is a versatile and popular high-level programming language!")

        # Keyword 3: 'jetson'
        elif "jetson" in user_input:
            print("Bot: NVIDIA Jetson is a series of embedded computing boards designed for AI at the edge.")

        # Keyword 4: 'ai'
        elif "ai" in user_input:
            print("Bot: Artificial Intelligence (AI) enables machines to simulate human intelligence and solve complex problems.")

        # Keyword 5: 'name'
        elif "name" in user_input:
            print("Bot: My name is QA Bot, your automated assistant!")

        # Default response when no keywords are matched
        else:
            print("Bot: I'm sorry, I don't understand that question yet. Try asking about 'Python', 'Jetson', 'AI', 'hello', or my 'name'.")

        print()  # Print an empty line for better readability

if __name__ == "__main__":
    qa_bot()
```
